```python
# Filter only Technology products
tech_df = df[df['Category'] == 'Technology']
print(tech_df.shape)
```

```python
# Filter Orders from the West region
west_df = df[df['Region'] == 'West']
print (west_df.shape)
```

```python
# Orders with Sales greater than 1000
high_sales = df[df['Sales'] > 1000]
print(high_sales.shape)
```

```python
# High value technology orders in the West
west_tech_high = df[
    (df['Region'] == 'West') &
    (df['Category'] == 'Technology') &
    (df['Sales'] > 1000)
    ]
print(west_tech_high.shape)
west_tech_high.head()
```

```python
# Top 10 highest Sales
df.sort_values('Sales',ascending=False).head(10)
```

```python
# Sort by Region then by Sales
df.sort_values(['Region','Sales'],ascending=[True,False]).head(10)
```

```python
# Total sales by region
region_sales = df.groupby('Region')['Sales'].sum().reset_index()
region_sales = region_sales.sort_values('Sales', ascending=False)
print(region_sales)
```

```python
# Total sales by product category
category_sales = df.groupby('Category')['Sales'].sum().reset_index()
category_sales = category_sales.sort_values('Sales', ascending=False)
print(category_sales)
```

```python
# Average order value by customer segment
segment_sales = df.groupby('Segment')['Sales'].mean().reset_index()
segment_sales = segment_sales.sort_values('Sales', ascending=False)
print(segment_sales)
```

```python
# Top 10 customers by total sales
top_customers = df.groupby('Customer Name')['Sales'].sum().reset_index()
top_customers = top_customers.sort_values('Sales', ascending=False).head(10)
print(top_customers)
```

```python
# Import visualisation library
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
# Bar chart of total sales by region
plt.figure(figsize=(7, 4))
plt.bar (x = region_sales['Region'], 
         height = region_sales['Sales'], 
         color='midnightblue')
plt.xticks(fontsize = 12)
plt.yticks(fontsize = 12)
plt.title('Total Sales by Region', fontsize = 16, fontweight = 'bold')
plt.xlabel('Region', fontsize = 13, weight = 'bold')
plt.ylabel('Total Sales ($)', fontsize = 13, weight = 'bold')
plt.savefig('sales_by_region.png')
plt.show()
```

```python
# Bar chart of total sales by category
plt.figure(figsize=(7, 4))
plt.bar(x = category_sales['Category'], 
        height = category_sales['Sales'], 
        color='coral')
plt.xticks(fontsize = 12)
plt.yticks(fontsize = 12)
plt.title('Total Sales by Category', fontsize = 16, fontweight = 'bold')
plt.xlabel('Category', fontsize = 13, weight = 'bold')
plt.ylabel('Total Sales ($)', fontsize = 13, weight = 'bold')
plt.savefig('sales_by_category.png')
plt.show()
```

```python
sns.set_palette('colorblind')
```

```python
# Pie chart of sales by customer segment
segment_total = df.groupby('Segment')['Sales'].sum()

plt.figure(figsize=(10, 5))
plt.pie(segment_total, labels=segment_total.index, 
        autopct='%.2f%%', 
        textprops = {'size' : 'x-large'},
                     colors=['steelblue', 'coral', 'green'])
plt.title('Sales Distribution by Customer Segment', fontsize = 18, weight = 'bold')
plt.savefig('sales_by_segment.png')
plt.show()
```

```python
# Convert Order Date to datetime with error handling
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, errors='coerce')
df['Month'] = df['Order Date'].dt.to_period('M')

# Group by Month and Category
area_data = df.groupby(['Month', 'Category'])['Sales'].sum().unstack()
sns.set_style('white')
plt.figure(figsize=(12, 6))
plt.stackplot(area_data.index.astype(str), 
              area_data['Furniture'], 
              area_data['Office Supplies'], 
              area_data['Technology'],
              labels=['Furniture', 'Office Supplies', 'Technology'],
              colors=['steelblue', 'coral', 'green'],
              alpha=0.8)
plt.title('Monthly Sales Trend by Category', fontsize = 16, weight = 'bold')
plt.xlabel('Month', fontsize = 13, fontweight = 'bold')
plt.ylabel('Total Sales ($)', fontsize = 13, fontweight = 'bold')
plt.xticks(rotation=90)
plt.legend(loc='upper left')
sns.despine()
plt.savefig('stacked_area_sales.png')
plt.show()
```
